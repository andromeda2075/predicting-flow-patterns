```python
model=make_model()
M='/home/guiomar/Desktop/CODES/predicting-flow-patterns/RESULTADOS PRELIMINARES/Unet_ConvLSTM_weights-20250721T013318Z-1-001/Unet_ConvLSTM_weights/Test3_Best_weights.weights.h5'
model.load_weights(M)
```

```python
xtrue=[]
ytrue=[]
N_batches = len(test_ds)
for k in range(N_batches):
    x,y=test_ds[k]
    xtrue.append(x)
    ytrue.append(x)
```

```python
# Parameters
warmup_runs = 5
batch_size = xtrue[0].shape[0]
repeat_count = 30

@tf.function
def infer(input_tensor):
    return model(input_tensor, training=False)
```

```python
# Warm-up con el primer batch
x_warmup = tf.convert_to_tensor(xtrue[0], dtype=tf.float32)
for _ in range(warmup_runs):
    _ = infer(x_warmup)

# Medición
tiempos_batch_promedio = []
tiempos_muestra_promedio = []

for i, x_batch in enumerate(xtrue):
        x_tensor = tf.convert_to_tensor(x_batch, dtype=tf.float32)
        tiempos_repetidos = []
        for _ in range(repeat_count):
            start = time.time()
            _ = infer(x_tensor)
            end = time.time()
            tiempos_repetidos.append((end - start) * 1000)  # milisegundos

        tiempo_promedio_batch = np.mean(tiempos_repetidos)
        tiempo_promedio_muestra = tiempo_promedio_batch / batch_size

        tiempos_batch_promedio.append(tiempo_promedio_batch)
        tiempos_muestra_promedio.append(tiempo_promedio_muestra)

        print(f"🧪 Batch {i+1}/{N_batches}: {tiempo_promedio_batch:.2f} ms ({tiempo_promedio_muestra:.2f} ms/muestra)")
       

# Estadísticas finales
print("\n📊 Estadísticas globales sobre los 200 batches:")
print(f"➡️ Tiempo medio por batch: {np.mean(tiempos_batch_promedio):.2f} ms")
print(f"➡️ Tiempo medio por muestra: {np.mean(tiempos_muestra_promedio):.2f} ms")
print(f"➡️ Desviación std por muestra: {np.std(tiempos_muestra_promedio):.2f} ms")
print(f"➡️ Mínimo por muestra: {np.min(tiempos_muestra_promedio):.2f} ms")
print(f"➡️ Máximo por muestra: {np.max(tiempos_muestra_promedio):.2f} ms")

```
